```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
import numpy as np
```

```python
data = []
classes = ['Anomaly', 'Normal']

for cls in classes:
    path = os.path.join("dataset_classificacao", cls)
    for img_name in os.listdir(path):
        img_path = os.path.join(path, img_name)
        try:
            with Image.open(img_path) as img:
                width, height = img.size
                img_array = np.array(img.convert('L'))  # Convert to grayscale
                mean_brightness = img_array.mean()
                data.append({
                        'filename': img_name,
                        'class': cls,
                        'width': width,
                        'height': height,
                        'brightness': mean_brightness,
                        'aspect_ratio': width / height
                    })
        except Exception as e:
            print(f"Error processing {img_path}: {e}")
            continue

```

```python
df = pd.DataFrame(data)
```

```python
df.info()
```

```python
# --- VISUALIZAÇÃO ---
plt.figure(figsize=(15, 10))

# 1. Distribuição de Classes
plt.subplot(2, 2, 1)
sns.countplot(data=df, x='class', palette='viridis')
plt.title('Equilíbrio de Classes (Anomaly vs Normal)')

# 2. Distribuição de Brilho
plt.subplot(2, 2, 2)
sns.histplot(data=df, x='brightness', hue='class', kde=True, element="step")
plt.title('Distribuição de Brilho Médio')

# 3. Dispersão de Resolução (Width vs Height)
plt.subplot(2, 2, 3)
sns.scatterplot(data=df, x='width', y='height', hue='class', alpha=0.5)
plt.title('Dimensões das Imagens (Resolução)')

# 4. Aspect Ratio
plt.subplot(2, 2, 4)
sns.boxplot(data=df, x='class', y='aspect_ratio')
plt.title('Proporção da Imagem (Aspect Ratio)')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 5, figsize=(20, 8))

for i, cls in enumerate(classes):
    path = os.path.join('dataset_classificacao', cls)
    imgs = os.listdir(path)[:5]  # Pega as 5 primeiras
    for j, img_name in enumerate(imgs):
        img = Image.open(os.path.join(path, img_name))
        axes[i, j].imshow(img)
        axes[i, j].set_title(f"{cls}\n{img.size}")
        axes[i, j].axis('off')

plt.suptitle("Amostras do Dataset", fontsize=16)
plt.show()
```
